```python
from tensorflow.keras.models import load_model

rede = load_model('rede_neural_tensor_04.h5')

from tensorflow.keras.preprocessing.image import load_img, img_to_array

imagem = load_img("WhatsApp Image 2025-02-21 at 07.40.17.jpeg",target_size=(128,128))

import matplotlib.pyplot as plt

plt.imshow(imagem)

imagem_array = img_to_array(imagem)/255

print(imagem_array)
import numpy as np

imagem_ajustada = np.expand_dims(imagem_array, axis=0)

previsao = rede.predict(imagem_ajustada)
print(previsao)
if(previsao[0][0] > 0.5):
    print("A Imagem processada é um cachorro!")
else:
    print("A Imagem processada é um gato!")

##TRANSFER LEARNING -- ESTUDAR
```
